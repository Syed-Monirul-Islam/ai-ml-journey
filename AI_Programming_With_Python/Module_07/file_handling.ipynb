```python
# read
with open ("./sample_data/test.txt", "r") as file:
   print(file.tell())
   print(file.read())

   print(file.tell())
```

```python
with open ("./sample_data/test.txt", "r") as file:
   print(file.tell())
   print(file.read(6))

   print(file.tell())
```

```python
# seek
with open ("./sample_data/test.txt", "r") as file:
   print(file.tell())
   print(file.read(6))

   print(file.tell())
   file.seek(0)
   print(file.tell())
```
