Question 1

```python
numbers = list(map(int, input().split()))

result = []
for num in numbers:
    if num % 4 == 0 and num % 8 != 0:
        result.append(num)

print(result)
```

question 2

```python
def compute_ratio(x, y):
    try:
        return (x + y) / (x * y)
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed."


x, y = map(float, input().split())

result = compute_ratio(x, y)
print(result)
```

question 4

```python
sentence = input()
words = sentence.split()

vowels = ("a", "e", "i", "o", "u")
result = []

for word in words:
    if word[0].lower() in vowels:
        result.append(word)

print(result)
```

question 5

```python
word = input().lower()

left = 0
right = len(word) - 1
is_palindrome = True

while left < right:
    if word[left] != word[right]:
        is_palindrome = False
        break
    left += 1
    right -= 1

if is_palindrome:
    print("Palindrome")
else:
    print("Not Palindrome")
```
