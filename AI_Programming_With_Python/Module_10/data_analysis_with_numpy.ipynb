# Ndarry

```python
import numpy as np

# 1d array

arr1 = np.array([10,20,30,40,50])

print(type(arr1))
print(arr1)

```

```python
#2d array
arr2 = np.array([[10,20,30], # 1st row
                 [60,70,80],  # 2nd row
                 [1,2,3]      # 3rd row
                ])

print(arr2)
print(arr2.shape)
```

```python
#3d array
arr3 = np.array([
                [[7,8,9,10],
                 [11,12,13,14]],

                [[10,20,30,40],
                 [70,80,90,100]]

])
print(arr3)
print(arr3.ndim)
print(arr3.shape)
print(arr3.size)
```

## array creation with random value

```python
# np.random.rand(shape)

arr =np.random.rand(2,3) # 0 -> 1
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.ndim)
print(arr.shape)
print(arr.size)
```

```python
# np.random.randint(start,end,shape)

arr =np.random.randint(1,10,(4,3))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.ndim)
print(arr.shape)
print(arr.size)
```

```python
# np.random.uniform(start,end,shape)

arr =np.random.uniform(10,20,(2,4,3))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.ndim)
print(arr.shape)
print(arr.size)
```

## array creation with range funtion

```python
# np.arange(start,end,step)

arr =np.arange(1,10,1)
mat =arr.reshape(3,3)

print(arr)
print(mat)

print(type(arr))
print(arr.dtype)
print(arr.ndim)
print(arr.shape)
print(arr.size)
```

```python
# np.linespace(start, end(inclued),  step)

arr =np.linspace(1,10,7)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.ndim)
print(arr.shape)
print(arr.size)
```
