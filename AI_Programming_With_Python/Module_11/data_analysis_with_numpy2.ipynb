###### seraching

```python
# np.where(condition , x, y) if true x, otherwise replace y
x = np.array([10, 20, 30, 40])

index = np.where(x==30)
print(index)

index = np.where(x > 30, x, 0)
print(index)
```

```python
mat = np.array([[10, 3, 5],
                [8, 4, 9]
                ])
print(mat)
```

```python
index =np.where( mat >8)
print(index)
mat2 =np.where( mat >8, mat, 0)
print(mat2)
```

## statistical function

```python
# data load

data = np.genfromtxt('student_scores.csv', delimiter=',', skip_header=1)

print(data)
```

```python
# statistical

math_marks = data[::, :1]
print(math_marks.T)

max_math_marks = np.max(math_marks)
print(max_math_marks)

min_math_marks = np.min(math_marks)
print(min_math_marks)

mean_math_marks = np.mean(math_marks)
print(mean_math_marks)

median_math_marks = np.median(math_marks)
print(median_math_marks)

# standard deviation
std_math_marks = np.std(math_marks)
print(std_math_marks)
```

## linear algebra

```python
A=np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B= np.array([
    [7, 8],
    [9, 10],
    [11,12]
])
# column of A = row of B
dot_product =np.dot(A,B)
print(dot_product)

# trace
print(np.trace(B))


```

```python
sq_mat = np.array([
          [15, 12, 4],
          [10, 5, 6],
          [17, 8, 9]
          ])
rank_sq=np.linalg.matrix_rank(sq_mat)
print(rank_sq)

det_of_sq = np.linalg.det(sq_mat)
print(det_of_sq)
```
