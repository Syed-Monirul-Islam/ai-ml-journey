```python
# You must run this code and use this dataset for all questions

import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

Q2

```python
max_salary = np.max(emp_income)
min_salary = np.min(emp_income)

avg_salary = np.mean(emp_income)
avg_age = np.mean(emp_ages)

# Print all results
print("Maximum Salary:", max_salary)
print("Minimum Salary:", min_salary)
print(f"Average Salary: {avg_salary:.2f}")
print(f"Average Age: {avg_age:.2f}")
```

Q3

```python
condition = (emp_ages >= 25) & (emp_ages <= 40) & (emp_income > 50000)

filtered_ages = emp_ages[condition]
filtered_salaries = emp_income[condition]
filtered_ids = emp_ids[condition]

print("--- Filtered Employees ---")
print("Employee IDs :", filtered_ids)
print("Ages         :", filtered_ages)
print("Salaries     :", filtered_salaries)


count = np.sum(condition)
print("Number of employees satisfying the condition:", count)
```

Q4

```python
avg_age = np.mean(emp_ages)

updated_ages = np.where(emp_ages < avg_age, emp_ages + 4, emp_ages + 2)

print(f"Average age used for condition: {avg_age:.2f}\n")
print("Original Array (emp_ages):")
print(emp_ages)
print()

print("Updated Array:")
print(updated_ages)
```

Q5

```python
tech_roles = ["Software Developer", "Backend Developer"]
is_tech = np.isin(emp_roles, tech_roles)

new_emp_income = np.where(is_tech, emp_income * 1.08, emp_income)

old_total_expense = np.sum(emp_income)
new_total_expense = np.sum(new_emp_income)
overall_increase = new_total_expense - old_total_expense

print(f"Old Total Expense     : {old_total_expense:,.2f}")
print(f"New Total Expense     : {new_total_expense:,.2f}")
print(f"Overall Increase      : {overall_increase:,.2f}")
```

```python

df["previous_salary"] = df["emp_salary"]

dev_roles = ["Software Developer", "Backend Developer"]
is_dev = df["role"].isin(dev_roles)


df.loc[is_dev, "new_salary"] = df.loc[is_dev, "previous_salary"] * 1.12

df.loc[~is_dev, "new_salary"] = df.loc[~is_dev, "previous_salary"] * 1.05

df["increase_of_salary"] = df["new_salary"] - df["previous_salary"]

result_df = df[["role", "previous_salary", "new_salary", "increase_of_salary"]]
display(result_df)
```
